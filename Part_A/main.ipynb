```python
class Intern:
    def __init__(self, intern_id, name, email, domain, duration):
        self.intern_id = intern_id
        self.name = name
        self.email = email
        self.domain = domain
        self.duration = duration

    def set_intern_id(self, intern_id):
        self.intern_id = intern_id

    def set_name(self, name):
        self.name = name

    def set_email(self, email):
        self.email = email

    def set_domain(self, domain):
        self.domain = domain

    def set_duration(self, duration):
        self.duration = duration

    def get_intern_id(self):
        return self.intern_id

    def get_name(self):
        return self.name

    def get_email(self):
        return self.email

    def get_domain(self):
        return self.domain

    def get_duration(self):
        return self.duration

    def __str__ (self):
        return f"intern_id: {self.intern_id}, name:{self.name}, email:{self.email}, domain:{self.domain}, duration:{self.duration}"
```

```python
from datetime import datetime

def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Function name is: ",func.__name__)
        t1 = datetime.now()
        res = func(*args, **kwargs)
        t2 = datetime.now()
        print("Execution time is: ",t2-t1)
        print("Result is: ",res)
        
        return res
    return wrapper

@my_decorator
def intern_performance_calculator(project, attendance):
    return (project*0.5) + (attendance*0.5)

intern_performance_calculator(9, 7)
```

```python
class Person:
    def __init__(self, name):
        self.name = name
    def about(self):
        print("name is:",self.name)

class Employee(Person):
    def __init__(self, name, empId):
        Person.__init__(self, name)
        self.empId = empId
    
    def about(self):
        print(f"name is:{self.name} and employee id is:{self.empId}")

class Mentor(Person):
    def __init__(self, name, mntId):
        Person.__init__(self, name)
        self.mntId = mntId

    def about(self):
        print(f"name is:{self.name} and mentor id is:{self.mntId}")

class TESRECOMentor(Employee, Mentor):
    def __init__(self, name, empId, mntId, domain):
        Employee.__init__(self, name, empId)
        Mentor.__init__(self, name, mntId)
        self.domain = domain

    def about(self):
        print(f"name is:{self.name} and employee id is:{self.empId} and mentor id is:{self.mntId} and domain is:{self.domain}")


TESMentor1 = TESRECOMentor("XYZ", "789", "123", "AI/ML")
TESMentor1.about()

mentor1 = Mentor("PQR", "123")
mentor1.about()

p1 = Person("ABC")
p1.about()

```

```python
from abc import ABC, abstractmethod

class Report(ABC):
    def __init__(self):
        print("instance of report abstract class is created")

    @abstractmethod
    def generate_report(self):
        pass

class AttendanceReport(Report):
    def __init__(self, attendance, total_days):
        self.attendance = attendance
        self.total_days = total_days

    def generate_report(self):
        print("your attendance report:",(self.attendance/self.total_days)*100)

class PerformanceReport(Report):
    def __init__(self, project_status, total_projects):
        self.project_status = project_status
        self.total_projects = total_projects

    def generate_report(self):
        print("your performance report:",(self.project_status/self.total_projects)*100)

attendance_report_1 = AttendanceReport(80, 100)
attendance_report_1.generate_report()

performance_report_1 = PerformanceReport(3, 4)
performance_report_1.generate_report()
```

```python
from functools import reduce

intern_score = [78, 90, 65, 88, 95]

def square(numbers):
    return numbers*numbers

map_used = list(map(square, intern_score))
print(map_used)

def even(num):
    if num%2==0:
        return True
    
filter_used = list(filter(even, intern_score))
print(filter_used)

reduce_used = reduce(lambda x, y: x+y, intern_score)
print(reduce_used)
```

```python
import csv

def add_record(name, email, duration, domain):
    with open("intern_details.csv", "a", newline="") as file:
        writer = csv.writer(file)
        writer.writerow([name, email, duration, domain])
    print("added successfully")

# add_record("jhon doe","jhon@gmail.com",4,"Devops")

def search_record(email):
    with open("intern_details.csv", "r") as file:
        reader = csv.reader(file)
        for row in reader:
            if row[1] == email:
                return row
    print("no detail found with this email")

# search_record("jhon@gmail.com")

def delete_record(email):
    rows = []
    found  = False

    with open("intern_details.csv", "r") as file:
        reader = csv.reader(file)
        for row in reader:
            if row[1] != email:
                rows.append(row)
            else:
                found = True
    
    with open("intern_details.csv", "w") as file:
        writer = csv.writer(file)
        writer.writerows(rows)
    
    if found:
        print("deleted successfully")
    else:
        print("no detail found with this email")

delete_record("jhon@gmail.com")
```

```python
import threading
import time

def attendance_processing():
    for i in range(1, 6):
        print(f"[Attendance] Processing intern {i}")
        time.sleep(1)


def certificate_generation():
    for name in ["ABC", "PQR", "XYZ", "JHON", "BOB"]:
        print(f"[Certificate] Generated for {name}")
        time.sleep(1.5)


t1 = threading.Thread(target=attendance_processing)
t2 = threading.Thread(target=certificate_generation)

t1.start()
t2.start()

t1.join()
t2.join()

print("Both tasks completed successfully.")
```

```python
from multiprocessing import Process
import time

def generate_performance_report(intern_name):
    print(f"Generating report for {intern_name}...")
    time.sleep(2)
    print(f"Performance report generated for {intern_name}")

if __name__ == "__main__":
    interns = ["ABC", "PQR", "XYZ", "JHON", "BOB"]

    processes = []

    for intern in interns:
        p = Process(target=generate_performance_report, args=(intern,))
        processes.append(p)
        p.start()

    for p in processes:
        p.join()

    # print(processes)

    print("All performance reports generated successfully.")
```
